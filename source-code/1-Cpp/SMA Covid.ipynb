```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as st
import pandas as pd
import csv
from statistics import mean,variance


NB_REPLICATION = 30


listOfListNbSusceptible =[]
listOfListNbExposed =[]
listOfListNbInfected =[]
listOfListNbRecovered =[] 



for replication in range(0,NB_REPLICATION):

    listNumIteration = []
    listNbSusceptible = []
    listNbExposed = []
    listNbInfected = []
    listNbRecovered = []


    f = open("log"+str(replication)+".txt", "r")
    fileLines = f.readlines()
    i = 0
    for line in fileLines:
        if("##" not in line):
            if(i == 0):
                listNumIteration.append(int(line))
                i = i + 1
            elif(i==1):
                listNbSusceptible.append(int(line))
                i = i + 1
            elif(i==2):
                listNbExposed.append(int(line))
                i = i + 1
            elif(i==3):
                listNbInfected.append(int(line))
                i = i + 1
            elif(i==4):
                listNbRecovered.append(int(line))
                i = i + 1
        else:
            i = 0

    f.close()



    nbHumainsTotal = listNbSusceptible[0] + listNbInfected[0]
    print(nbHumainsTotal)

    listNbSusceptible = [i/nbHumainsTotal for i in listNbSusceptible]
    listNbExposed = [i/nbHumainsTotal for i in listNbExposed]
    listNbInfected = [i/nbHumainsTotal for i in listNbInfected]
    listNbRecovered = [i/nbHumainsTotal for i in listNbRecovered]

    listOfListNbSusceptible.append(listNbSusceptible)
    listOfListNbExposed.append(listNbExposed)
    listOfListNbInfected.append(listNbInfected)
    listOfListNbRecovered.append(listNbRecovered)
    
    plt.figure(str(replication))
    print("Courbe des résultats sur la réplication num : " +str(replication))
    plt.plot(listNumIteration, listNbSusceptible, label="susceptible")  # Matplotlib plot.
    plt.plot(listNumIteration, listNbExposed, label="exposed")  # Matplotlib plot.
    plt.plot(listNumIteration, listNbInfected, label="infected")  # Matplotlib plot.
    plt.plot(listNumIteration, listNbRecovered, label="recovered")  # Matplotlib plot.
    plt.xlabel('Time in days')  # Add an x-label to the axes.
    plt.ylabel('Ratio of agents in each state')  # Add a y-label to the axes.
    plt.title("S, E, I and R agents population over time")  # Add a title to the axes.
    plt.legend(loc="best")
    plt.savefig("resultats_replication"+str(replication)+".png")
    
    """
    string = "resultats_replication"+str(replication)+".csv"
    with open(string, "w") as fCsv:
        write = csv.writer(fCsv)
        write.writerow(listNbSusceptible)
        write.writerow(listNbExposed)
        write.writerow(listNbInfected)
        write.writerow(listNbRecovered)
        
    """
    f = open("ResultatRepli"+str(replication)+".csv", "w")
    for num in listNumIteration:
        f.write(str(listNbSusceptible[num]))
        f.write(",")
        f.write(str(listNbExposed[num]))
        f.write(",")
        f.write(str(listNbInfected[num]))
        f.write(",")
        f.write(str(listNbRecovered[num]))
        f.write("\n")
            
        
    f.close()
```

```python
avgNbSusceptible = [mean(value) for value in zip(*listOfListNbSusceptible)]
avgNbExposed = [mean(value) for value in zip(*listOfListNbExposed)]
avgNbInfected = [mean(value) for value in zip(*listOfListNbInfected)]
avgNbRecovered = [mean(value) for value in zip(*listOfListNbRecovered)] 

print(avgNbSusceptible)
```

```python
plt.figure("moyenne")
print("Courbe des résultats sur la moyenne")
plt.plot(listNumIteration, avgNbSusceptible, label="susceptible")  # Matplotlib plot.
plt.plot(listNumIteration, avgNbExposed, label="exposed")  # Matplotlib plot.
plt.plot(listNumIteration, avgNbInfected, label="infected")  # Matplotlib plot.
plt.plot(listNumIteration, avgNbRecovered, label="recovered")  # Matplotlib plot.
plt.xlabel('Time in days')  # Add an x-label to the axes.
plt.ylabel('Ratio of agents in each state')  # Add a y-label to the axes.
plt.title("S, E, I and R agents population over time")  # Add a title to the axes.
plt.legend(loc="best")
plt.savefig("resultats_moyenne.png")
string = "resultats_moyenne.csv"
with open(string, "w") as fCsv:
    write = csv.writer(fCsv)
    write.writerow(avgNbSusceptible)
    write.writerow(avgNbExposed)
    write.writerow(avgNbInfected)
    write.writerow(avgNbRecovered)
```
