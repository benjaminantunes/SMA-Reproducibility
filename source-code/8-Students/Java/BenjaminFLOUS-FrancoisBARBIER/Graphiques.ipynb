```python
#Premier module pour afficher les 100 simulations.

import matplotlib.pyplot as plt

all_data = []

# Boucle sur 100 fichiers
for i in range(1, 101):
    filename = f'resultats/Resultat_Simulation_{i}.csv'  # Remplacez le nom du fichier par le format réel que vous utilisez

    # Lire les données à partir du fichier texte
    data = []
    with open(filename, 'r') as file:
        for line in file:
            fields = line.strip().split(' ')
            if len(fields) == 5:
                data.append([int(x) for x in fields])

    # Extraire les données du fichier
    nb_tour, nb_S, nb_E, nb_I, nb_R = zip(*data)
    
    # Diviser les données par 20000
    nb_S = [x / 20000 for x in nb_S]
    nb_E = [x / 20000 for x in nb_E]
    nb_I = [x / 20000 for x in nb_I]
    nb_R = [x / 20000 for x in nb_R]

    # Créer une nouvelle fenêtre pour chaque graphique
    plt.figure()

    # Tracer le graphique sur la fenêtre actuelle
    plt.plot(nb_tour, nb_S, label='Susceptible')
    plt.plot(nb_tour, nb_E, label='Exposé')
    plt.plot(nb_tour, nb_I, label='Infecté')
    plt.plot(nb_tour, nb_R, label='Rétabli')

    # Ajouter des étiquettes et une légende
    plt.xlabel('Temps_en_jour')
    plt.ylabel('Population')
    plt.title(f'Population d_individus S, E, I et R dans le temps après la simulation {i}')
    plt.legend()
    
    # Enregistrer le graphique dans un fichier
    plt.savefig(f'graphiques/Simulation_{i}.png')

    # Fermer la fenêtre de graphique actuelle
    plt.close()
```

```python
#Second module pour afficher la moyenne des 100 simulations.

import matplotlib.pyplot as plt
import numpy as np

# Créer des tableaux vides pour stocker les données de toutes les simulations
all_data = []

# Boucle sur 100 fichiers
for i in range(1, 101):
    filename = f'resultats/Resultat_Simulation_{i}.csv'  # Remplacez le nom du fichier par le format réel que vous utilisez

    # Lire les données à partir du fichier texte
    data = []
    with open(filename, 'r') as file:
        for line in file:
            fields = line.strip().split(' ')
            if len(fields) == 5:
                data.append([int(x) for x in fields])

    # Ajouter les données de cette simulation à la liste
    all_data.append(data)

# Calculer la moyenne des données de toutes les simulations
average_data = np.mean(all_data, axis=0)

# Extraire les données du fichier
nb_tour, nb_S, nb_E, nb_I, nb_R = zip(*average_data)

# Diviser les données par 20000
nb_S = [x / 20000 for x in nb_S]
nb_E = [x / 20000 for x in nb_E]
nb_I = [x / 20000 for x in nb_I]
nb_R = [x / 20000 for x in nb_R]

# Créer un graphique pour afficher la moyenne
plt.figure()

# Tracer le graphique de la moyenne
plt.plot(nb_tour, nb_S, label='Moyenne Susceptible')
plt.plot(nb_tour, nb_E, label='Moyenne Exposé')
plt.plot(nb_tour, nb_I, label='Moyenne Infecté')
plt.plot(nb_tour, nb_R, label='Moyenne Rétabli')

# Ajouter des étiquettes et une légende
plt.xlabel('Temps_en_jour')
plt.ylabel('Population')
plt.title('Moyenne de la Population S, E, I et R dans le temps')
plt.legend()

# Enregistrer le graphique de la moyenne
plt.savefig('Moyenne_Population.png')

# Afficher le graphique de la moyenne
plt.show()
```
