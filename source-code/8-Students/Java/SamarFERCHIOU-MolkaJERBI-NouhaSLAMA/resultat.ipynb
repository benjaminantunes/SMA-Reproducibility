```python
import pandas as pd
import matplotlib.pyplot as plt

list_of_dfs = []
colors =["green", "orange", "blue", "red"]
x = range(730)

for i in range(100):
    df = pd.read_csv("resultats/exp"+str(i)+".csv")
    list_of_dfs.append(df)
```

Les courbes des 100 expériences

```python
j = 0
for df in list_of_dfs:
    fig, ax = plt.subplots()
    for i in range(4):
        y = df[df.columns[i]].tolist()
        plt.plot(x,y, label = df.columns[i], color = colors[i])
    ax.set_xlabel('Time in days')
    ax.set_ylabel('Ratio of agents in each state')
    ax.set_title('S, E, I and R agents population over time experiment '+str(j))
    
    ax.legend()
    plt.savefig("courbes/courbe"+str(j)+".png")
    j+=1
    
plt.show()
```

La courbe de la moyenne

```python
fig, ax = plt.subplots()

for i in range(4):
     y = []
     for j in range(730):
         moy = 0.0
         for df in list_of_dfs: 
             moy += float(df[df.columns[i]].tolist()[j])
         y.append(moy / 100)
     plt.plot(x,y, label = df.columns[i], color = colors[i])
    

ax.set_xlabel('Time in days')
ax.set_ylabel('Ratio of agents in each state')
ax.set_title('S, E, I and R agents population over time')

ax.legend()
plt.savefig("courbeMoyenne.png")

plt.show()
```
