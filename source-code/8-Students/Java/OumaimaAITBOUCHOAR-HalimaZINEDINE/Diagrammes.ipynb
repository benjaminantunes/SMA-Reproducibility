```python
import pandas as pd
import matplotlib.pyplot as plt

list = []
colors = ["blue","orange","green", "red"]
x = range(731)

for i in range(100):
    try:
        df = pd.read_csv("Simulations/Simulation" + str(i) + ".csv")
        list.append(df)
    except pd.errors.ParserError:
        print(f"Error reading file: Simulation_data{i}.csv. Skipping...")
        continue
```

```python
j = 0
for df in list:
    fig, ax = plt.subplots()
    for i in range(4):
        y = df[df.columns[i]].tolist()
        plt.plot(x,y, label = df.columns[i], color = colors[i])
    ax.set_xlabel('Time in days')
    ax.set_ylabel('Ratio of agents in each state')
    ax.set_title('S, E, I and R agents population over time '+str(j))
    
    ax.legend()
    plt.savefig("courbes/courbe"+str(j)+".png")
    j+=1
    
plt.show()
```

```python
 ig, ax = plt.subplots()

for i in range(4):
     y = []
     for j in range(731):
         moyenne = 0.0
         for df in list: 
             moyenne += float(df[df.columns[i]].tolist()[j])
         y.append(moyenne / 100)
     plt.plot(x,y, label = df.columns[i], color = colors[i])
    

ax.set_xlabel('Time in days')
ax.set_ylabel('Ratio of agents in each state')
ax.set_title('S, E, I and R agents population over time')

ax.legend()
plt.savefig("LaCourbeMoyenne.png")

plt.show()
```
