# Modèle Toy case SEIR multi-agent

```python
import matplotlib.pyplot as plt
import pandas as pd
import os

folder_name = "simulation/"

list_dataframes = []

for file in os.listdir(folder_name):
    if file.endswith(".csv"):
        path = os.path.join(folder_name, file)
        
        dataframe = pd.read_csv(path)
        dataframe = dataframe.drop(dataframe.columns[0], axis=1)

        list_dataframes.append(dataframe)
```

```python
for i, dataframe in enumerate(list_dataframes):
    print(f"\nDataFrame {i + 1}:\n")
    print(dataframe)
```

```python
add_data = list_dataframes[0]

for i, dataframe in enumerate(list_dataframes[1:], start=1):
    add_data = add_data.add(dataframe, fill_value=0)

print(add_data)
```

```python
mean_data = add_data / len(list_dataframes)

mean_data
```

```python
for simu in list_dataframes:
    simu.plot()
    plt.ylabel('Ratio of agents in each state')
    plt.title('S, E, I, R agents population over time')
    plt.legend()

    # Afficher le graphique
    plt.show()
```

```python
# Utiliser la première ligne comme référence pour le ratio
df_ratio = mean_data.div(mean_data.iloc[0].sum())

df_ratio
```

```python
plt.figure(figsize=(10, 6))

# Parcourir chaque colonne et tracer la ligne correspondante
for col in df_ratio.columns:
    plt.plot(df_ratio.index, df_ratio[col], label=col)

# Ajouter des labels et une légende
plt.xlabel('Time in days')
plt.ylabel('Ratio of agents in each state')
plt.title('S, E, I, R agents population over time')
plt.legend()

# Afficher le graphique
plt.show()
```
