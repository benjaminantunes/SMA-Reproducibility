```python
import pandas as pd
import matplotlib.pyplot as plt
import glob

# Créez une liste pour stocker les moyennes quotidiennes
moyenne_susceptibles = []
moyenne_exposes = []
moyenne_infectes = []
moyenne_recovered = []

# Utilisez glob pour obtenir la liste des fichiers CSV dans le répertoire
files = glob.glob("fichiersResultatsCSV/*.csv")  # Assurez-vous que les fichiers sont dans le même répertoire que votre script Python

# Parcourez chaque fichier CSV
for file in files:
    df = pd.read_csv(file, header=None, sep=";")  # Lire le fichier CSV
    susceptibles = df.iloc[:, 0].tolist()  # Accédez à la première colonne
    exposes = df.iloc[:, 1].tolist()  # Accédez à la deuxième colonne
    infectes = df.iloc[:, 2].tolist()  # Accédez à la troisième colonne
    recovered = df.iloc[:, 3].tolist()  # Accédez à la quatrième colonne
    
    # Ajoutez les valeurs quotidiennes aux listes de moyennes
    moyenne_susceptibles.append(susceptibles)
    moyenne_exposes.append(exposes)
    moyenne_infectes.append(infectes)
    moyenne_recovered.append(recovered)

# Calculez les moyennes quotidiennes
moyenne_susceptibles = [sum(x) / len(x) for x in zip(*moyenne_susceptibles)]
moyenne_exposes = [sum(x) / len(x) for x in zip(*moyenne_exposes)]
moyenne_infectes = [sum(x) / len(x) for x in zip(*moyenne_infectes)]
moyenne_recovered = [sum(x) / len(x) for x in zip(*moyenne_recovered)]

# Créez une liste de jours (0 à 730)
jours = list(range(730))

# Créez un graphique basé sur les moyennes quotidiennes
plt.figure(figsize=(12, 6))
plt.plot(jours, moyenne_susceptibles, label="Moyenne des Susceptibles", linestyle='-', marker='o')
plt.plot(jours, moyenne_exposes, label="Moyenne des Exposés", linestyle='-', marker='o')
plt.plot(jours, moyenne_infectes, label="Moyenne des Infectés", linestyle='-', marker='o')
plt.plot(jours, moyenne_recovered, label="Moyenne des Recovered", linestyle='-', marker='o')

# Ajoutez des légendes et un titre
plt.legend()
plt.title("Moyenne de l'évolution de la population sur 100 fichiers")
plt.xlabel("Jours")
plt.ylabel("Nombre moyen de personnes")

# Affichez le graphique
plt.grid(True)
plt.show()
```
